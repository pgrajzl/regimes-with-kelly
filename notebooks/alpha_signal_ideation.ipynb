<h3 style="text-align: center;">Alpha Signal Ideation: Rolling Beta vs. Future Excess Returns</h3>

___

*Hypothesis:* a stock's trailing (historical) rolling beta may contain predictive information about its *future* excess returns vs. SPY — i.e., does a stock's past market sensitivity say anything about how much it goes on to beat or lag the market afterward?

This is closely related to the well-documented "betting against beta" anomaly in academic finance (Frazzini & Pedersen), which finds that low-beta stocks have historically delivered higher risk-adjusted returns than high-beta stocks — the opposite of what CAPM would predict. Worth testing whether a simpler, more direct version of that relationship shows up in this data: does trailing beta level predict subsequent excess return, across the S&P 500 universe?
___

<h3 style="text-align: center;">The Core Question</h3>

___

For a given stock, at a given point in time:
- *Trailing beta* — computed over some lookback window (e.g. 6mo, 12mo) ending today
- *Forward excess return* — the stock's excess return vs. SPY over some future window (e.g. the next 1mo, 3mo, 6mo) starting today

Might be there be a systematic relationship between the two? This sounds like a relatively naive signal to test initially, but an interesting follow-up here (if initial results prove unfruitful) is comparing delta in trailing beta (across different time frames) compared to forward excess returns.

<h3 style="text-align: center;">Research Plan</h3>


##### *Core signal test*
- Compute rolling beta (trailing lookback) vs. forward excess return, per stock, per date
- IC = cross-sectional correlation (Spearman) between trailing beta and forward excess return, computed per date, then averaged across dates (standard IC methodology)
- Naive first pass: single lookback, single forward window, full S&P universe, no sector splits

##### *Required infrastructure*
- Sweep across multiple rolling beta lookback windows (e.g. 1mo, 3mo, 6mo, 12mo) — not hardcoded to one
- Sweep across multiple forward return horizons (e.g. 1mo, 3mo, 6mo) — same
- Sector tagging per ticker (from the S&P constituent list — Wikipedia's table already includes GICS sector, worth pulling that alongside tickers), so IC can be computed within-sector as well as pooled
- A generalized IC calculation function; not specific to beta, since factor sensitivities (value, momentum, quality, size, low-vol) will need the same IC machinery later, just with a different signal column swapped in
- A signal panel structure: long-format DataFrame (date, ticker, signal_value, forward_return, sector) is probably the right shape, since it supports groupby on any dimension (date for cross-sectional IC, sector for sector-conditioned IC, ticker for time-series checks) without reshaping every time

##### *Extensibility*
- Multiple signals computed and tested side by side (beta today, factor sensitivities later); same panel/IC infrastructure should handle both without a rewrite
- Signal ranking will inform long/short portfolio construction (top-N / bottom-N by signal, or full cross-sectional ranking with position weights)
- Kelly-based position sizing layered on top of the ranking
- Regime conditioning: namely, testing whether IC itself is regime-dependent (does trailing beta predict forward excess return well in some regimes and not others), which requires regime detection modeling as an input to this pipeline

##### *Risks and Considerations*
- Look-ahead bias: forward return window must start strictly after the beta lookback window ends, zero overlap
- Survivorship bias: today's S&P constituent list only
- Overfitting via sweeping: testing many lookback/horizon combinations increases the chance some combination looks good by chance; whatever combination looks best needs out-of-sample/walk-forward validation before it's trusted, not just in-sample IC
